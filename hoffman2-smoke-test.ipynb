# Hoffman2 machine learning smoke test

Fit a linear regression model on synthetic training data and measure its mean absolute error on a held-out test set. The test uses only Python's standard library and runs on a CPU.

In a Hoffman2 Jupyter session, the output also confirms whether Slurm provided a job ID.

```python
import os
from random import Random
from statistics import linear_regression, mean

rng = Random(42)
features = list(range(100))
rng.shuffle(features)
targets = [3 * x + 5 + rng.uniform(-1, 1) for x in features]

split = 80
slope, intercept = linear_regression(features[:split], targets[:split])
predictions = [slope * x + intercept for x in features[split:]]
test_mae = mean(abs(actual - predicted) for actual, predicted in zip(targets[split:], predictions))

print(f"Slurm job ID available: {bool(os.environ.get('SLURM_JOB_ID'))}")
print(f"Training samples: {split}, test samples: {len(predictions)}")
print(f"Fitted model: y = {slope:.2f}x + {intercept:.2f}")
print(f"Held-out mean absolute error: {test_mae:.3f}")
assert test_mae < 1.0, "Expected held-out mean absolute error below 1.0"
```
